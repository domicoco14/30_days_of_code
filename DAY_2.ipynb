# **Gender VS. Survival**

```python
print(titanic["Sex"].value_counts())
sns.barplot(data=titanic, x= 'Sex', y= 'Survived')
plt.show()
```

**Insight:**

- Women had a much higher survival rate

- Clear evidence that women were given priority during evacuation.

**Gender was one of the strongest predictors of survival.**

#

# **Age Range Vs. Survival**

```python
titanic["Age"].value_counts().sum()

bins = [0, 12, 19, 35, 100]
categories = ["Children", "Teenagers", "Youth", "Adults"]

titanic["Age_Range"] = pd.cut(titanic["Age"], bins=bins, labels=categories, right=True)
sns.barplot(data=titanic, x= 'Age_Range', y= 'Survived')
plt.show()
```

**Insight:**

- Children had the highest survival rate

- Survival rates decreased with increasing age

**Interpretation:**

**Age played a significant role in survival, with younger passengers being prioritized during evacuation, especially children.**

```python
titanic["Embarked"].value_counts()
sns.barplot(data=titanic, x= 'Embarked', y= 'Survived')
plt.show()
```

```python
# More people died than they Survived in the accident
print(titanic["Survived"].value_counts())
sns.histplot(data=titanic, x= 'Survived')
plt.show()
```

# **Passenger Class (P-Class) vs Survival**



```python
# First Class survived the titanic accident than second class or third class
titanic["Pclass"].value_counts()
sns.barplot(data=titanic, x= 'Pclass', y= 'Survived')
plt.show()
```

**Insight:**

- 1st class passengers survived far more

- 3rd class had the lowest survival rate

**Interpretation:**

Socioeconomic status played a major role in survival.

# **Family_Size vs. Survival**

```python
titanic["SibSp"].value_counts().sum()
titanic["Parch"].value_counts().sum()
traveller = 1
titanic["Family_Size"] = titanic["SibSp"] + titanic["Parch"] + traveller
sns.barplot(data=titanic, x= 'Family_Size', y= 'Survived')
plt.show()
```

**Insight:**

- Passengers with medium family sizes (2–4) had the highest survival rates

- Solo travelers (Family Size = 1) and very large families (5+) had lower survival rates

**Interpretation:**

Having a small family group increased chances of survival, while traveling alone or in very large groups reduced survival likelihood due to limited support or difficulty coordinating during evacuation.

# **Fare Vs Survival**

```python
# Passengers who survived tend to have paid higher fares on average.
titanic["Fare"]
sns.barplot(data=titanic, x="Survived", y= "Fare")
plt.show()
```

**Insights:**

- Survivors paid higher fares on average

- Fare has outliers (some tickets were extremely expensive)

- Fare is right-skewed

# **Correlation between Variables**
