```python
import pandas as pd

titanic_feature = pd.read_csv('/content/drive/MyDrive/30_days_of_Code/titanic_feature.csv')

# Display the first 5 rows of the DataFrame
print(titanic_feature.head(3))
```

```python
titanic_feature['Age_Range'].value_counts()
```

```python
sex_dummies = pd.get_dummies(
    titanic_feature['Sex'],
    prefix='Sex',
    dtype=int
)
```

```python
titanic_feature = pd.concat(
    [titanic_feature, sex_dummies],
    axis=1
)
titanic_feature.drop(columns=['Sex'], inplace=True)
titanic_feature.columns
```

```python
titanic_feature = titanic_feature.loc[:, ~titanic_feature.columns.duplicated()]
titanic_feature.columns
```

```python
titanic_feature["Age"].value_counts().sum()

bins = [0, 12, 19, 35, 100]
categories = ["Children", "Teenagers", "Youth", "Adults"]

titanic_feature["Age_Range"] = pd.cut(titanic_feature["Age"], bins=bins, labels=categories, right=True)
```

```python
titanic_feature = titanic_feature.loc[
    :, ~titanic_feature.columns.str.startswith('Age_')
]
```

```python
titanic_feature.columns
```

```python
age_dummies = pd.get_dummies(
    titanic_feature['Age_Range'],
    prefix='Age',
    dtype=int
)

titanic_feature = pd.concat(
    [titanic_feature, age_dummies],
    axis=1
)
```

```python
titanic_feature.drop(columns=['Age_Range'], inplace=True)
```

```python
titanic_feature["Embarked"] = titanic["Embarked"]
```

```python
embarked_dummies = pd.get_dummies(titanic_feature["Embarked"], prefix="Embarked", dtype=int)
titanic_feature = pd.concat([titanic_feature, embarked_dummies], axis=1)
titanic_feature.drop(columns=["Embarked"], inplace=True)
titanic_feature.columns
```

```python
def drop_unneeded_columns(df):
    """
    Drop unnecessary columns from the Titanic dataset.
    """
    cols_to_drop = ['PassengerId', 'Name', 'Ticket', 'Cabin', 'Title','Age_Range']
    return df.drop(columns=[col for col in cols_to_drop if col in df.columns])

titanic_feature = drop_unneeded_columns(titanic_feature)
titanic_feature.columns
```

```python
features = ['Pclass', 'Age', 'SibSp', 'Parch', 'Fare', 'Family Size',
       'IsAlone', 'IsChild', 'Age_Children', 'Age_Teenagers', 'Age_Youth',
       'Age_Adults', 'Sex_female', 'Sex_male', 'Embarked_C',
       'Embarked_Q', 'Embarked_S']

target = 'Survived'

print("=== Features used for modeling: === ")
print(features)
print("\n=== Target variable: === ")
print(target)
```
