```python
import pandas as pd

titanic = pd.read_csv('/content/drive/MyDrive/30_days_of_Code/titanic.csv').copy()

# Writing a function to fill all missing values with the mean
def fill_na(df, column, sum_stats):
  """A function that takes a dataframe and its column then fill the missing value with its mean"""
  if sum_stats == 'mean':
    df[column] = df[column].fillna(df[column].mean())

  elif sum_stats == 'median':
    df[column] = df[column].fillna(df[column].median())

  elif sum_stats == 'mode':
    df[column] = df[column].fillna(df[column].mode()[0])

  else:
    df[column] = df[column].dropna()

  return df

titanic = fill_na(titanic, 'Age', 'mean')
titanic = fill_na(titanic, 'Cabin', 'mode')
titanic = fill_na(titanic, 'Embarked', 'mode')
```
